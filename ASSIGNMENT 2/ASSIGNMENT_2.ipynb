```python
#LIST CREATION, ACCESS AND MODIFICATION

# 1.LIST CREATION

age_list = [24, 25, 26, 27, 28]
print(age_list)
#The list contains five integer values and age_list is the variable storing the list.

name_list = ["Akash", "Arun", "Rahul", "Anu", "Meera"]
print(name_list)
#created a list named name_list with string elements inside "".

# 2.LIST OPERATIONS/MODIFICATIONS

name_list.append("Yazhini")
print(name_list)
#append() adds an element to the end of the list named name_list.

age_list.insert(2, 30)
print(age_list)
```

the index value in a list starts from 0 which inserts the value 30 to the 3rd position.
   2 indicates 2nd index position and 30 is the value to be added to the position


```python

name_list.remove("Yazhini")
print(name_list)
```

remove() removes a specific value from the list.
   It searches for "Yazhini" and removes it


```python

age_list.pop()
print(age_list)
#pop() removes the last element from a list when no index position is provided.

age_list.extend([29, 30, 26])
print(age_list)
```

append() would add the entire list as one element, whereas extend() adds each value separately.
also multiple values could be added to the list using extend()


```python

age_list.sort(reverse=True)
print(age_list)
```

normal sort() sorts the values from smallest to largest and 
sort(reverse=true) sorts the values from largest to smallest.


```python

print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
#max() finds maximum value, min() finds the minimum vale and sum() adds up the whole values in the list.

# 3.ACCESSING LIST ELEMENTS

print(name_list[0])
#print(name_list[0]).

print(name_list[-1])
#print(name_list[-1])
```

```python

print(name_list[2:5])
#Python slicing has the format list[start:end] and the end is excluded.

print(name_list[::-1])
#The slicing format is list[start:end:step].[::-1] means move through the list with a step of -1, so it reverses the list.

#DICTIONARY(CREATION,MODIFICATION AND ACCESS)

student_marks = {
    "Akash": 75,
    "Arun": 68,
    "Rahul": 90,
    "Anu": 82,
    "Meera": 95
}
print(student_marks)
#dictionary stores data as key : value.

print(student_marks["Rahul"])
#The key "Rahul" is used to access its corresponding value.

student_marks["Janani"] = 80
print(student_marks)
#If the key doesn't already exist, assigning a value creates a new dictionary entry.

student_marks["Akash"] = 82
print(student_marks)
#Because "Akash" already exists, Python updates its value instead of creating another "Akash" key.

print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
#keys() print all dictionary keys, values() print all dictionary values and items() print both dictionary keys and values.


#SETS

my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print(my_set)
```

set automatically removes duplicate values. Here 'a' and 'i' have duplicates so set automatically removes it.
Also set is unordered so the output order will not be same as input order.


```python

my_set[4] = 's'
#It shows error as set is an unordered collection of data and do not support indexing.

set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

union_set = set1.union(set2)
print("Union:", union_set)
#Union combines all elements from both sets and common values are not repeated.the output has unique values only.

intersection_set = set1.intersection(set2)
print("Intersection:", intersection_set)
#Intersection gives only the elements that are present in both sets.

#OPERATORS AND CONDITIONAL STATEMENTS(IF,ELIF,ELSE)

score = float(input("Enter your score (0 to 10): "))
if score > 7:
    print("Above Average: Excellent performance! Keep up the good work.")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```

If the score input is greater than 7 then the 1st condition output is printed.
if the 1st condition isn't true and the score is 4 or more, then the 2nd output is printed.
finally if neither condition satisfies then the 3rd output is given.
